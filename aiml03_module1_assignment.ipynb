# Question 1:

Write the code to convert an arbitrary string of 1s and 0s (binary) to hex

1. Write a function called bin_to_hex_digit() that takes up to 4 binary digits and converts it to 0-10, A, B, C, D, E, or F. Do NOT use build-int hex conversions, but a big if-then-else construct that tests the string. If the passed-in string is less than 4 digits, pad it in the front with '0' until it is length 4.

2. Write a function called bin_to_hex() that takes a string of 1s and 0s. It should loop backwards through the string, taking 4 digits at a time and call bin_to_hex_digit() to get the equavilent hex digit.

3. Each hex digit from step 2 should be concatented to the beginning of the string with the final answer, which is returned.








```python
def bin_to_hex_digit(binary_string):
    """
    Converts a binary string of up to 4 digits to a single hexadecimal digit.
    Pads the string with leading zeros if it is less than 4 digits long.
    """
    padded_string = binary_string.zfill(4)

    if padded_string == "0000":
        return "0"
    elif padded_string == "0001":
        return "1"
    elif padded_string == "0010":
        return "2"
    elif padded_string == "0011":
        return "3"
    elif padded_string == "0100":
        return "4"
    elif padded_string == "0101":
        return "5"
    elif padded_string == "0110":
        return "6"
    elif padded_string == "0111":
        return "7"
    elif padded_string == "1000":
        return "8"
    elif padded_string == "1001":
        return "9"
    elif padded_string == "1010":
        return "A"
    elif padded_string == "1011":
        return "B"
    elif padded_string == "1100":
        return "C"
    elif padded_string == "1101":
        return "D"
    elif padded_string == "1110":
        return "E"
    elif padded_string == "1111":
        return "F"


def bin_to_hex(binary_number_str):
    """
    Converts a binary number string to its hexadecimal equivalent.
    """
    hex_result = ""
    start = len(binary_number_str)

    while start > 0:
        end = start
        start = max(0, start - 4)

        chunk = binary_number_str[start:end]

        hex_digit = bin_to_hex_digit(chunk)
        hex_result = hex_digit + hex_result

    return hex_result
```

# Question 2

Write a function that counts the number of words in a sentence.

```python
def string_to_word_list(text):
    """
    Splits a string into a list of words.

    Args:
        text: The input string.

    Returns:
        A list of words from the input string.
    """
    word = ""
    word_list = []

    for char in text:
        if char != " ":
            word += char
        else:
            if word != "":
                clean_word = ""
                for c in word:
                    if c.isalnum() or c in ["'", "-"]:
                        clean_word += c

                if clean_word != "":
                    word_list.append(clean_word)

                word = ""

    if word != "":
        clean_word = ""
        for c in word:
            if c.isalnum() or c in ["'", "-"]:
                clean_word += c

        if clean_word != "":
            word_list.append(clean_word)

    return len(word_list)
```
